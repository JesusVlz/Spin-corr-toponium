```python
#!/usr/bin/env python3

# ============================================================
# COMPLETE SINGLE JUPYTERLAB CELL
#
# CMS DATA AND GENERATOR-LEVEL SPIN-CORRELATION OBSERVABLES
#
# Samples:
#   1. CMS Data
#   2. pp -> tstbar
#   3. Simplified EFT eta_j full
#   4. Simplified EFT eta_j restrict
#
# Each figure contains:
#   - normalized differential distribution
#   - Data/MC ratio panel
#   - PDF and PNG output
# ============================================================

import os
import uproot
import numpy as np
import matplotlib.pyplot as plt
import mplhep as hep

from matplotlib.lines import Line2D
from matplotlib.gridspec import GridSpec


# ============================================================
# STANDARD STYLE CONFIGURATION
# ============================================================

plt.style.use(hep.style.CMS)

USE_LATEX = True
plt.rcParams["text.usetex"] = USE_LATEX

FIGSIZE = (10, 10)
DPI = 300

LABEL_SIZE = 30
YLABEL_SIZE = 34
TITLE_SIZE = 28
LEGEND_SIZE = 20
TICK_SIZE = 24
RATIO_LABEL_SIZE = 25

LINE_WIDTH = 2.7
ERROR_LINE_WIDTH = 1.4
CAP_SIZE = 2.5

DATA_MARKER_SIZE = 7.0
RATIO_MARKER_SIZE = 5.5

plt.rcParams.update({
    "figure.figsize": FIGSIZE,
    "axes.labelsize": LABEL_SIZE,
    "xtick.labelsize": TICK_SIZE,
    "ytick.labelsize": TICK_SIZE,
    "legend.fontsize": LEGEND_SIZE,
    "lines.linewidth": LINE_WIDTH,
})


# ============================================================
# INPUT / OUTPUT CONFIGURATION
# ============================================================

DATA_DIR = "../jesus_github/Spin-corr-toponium/data/"
MC_DIR   = "../jesus_github/Spin-corr-toponium/outputs/"
OUTDIR   = "plots_style_ratio"

os.makedirs(OUTDIR, exist_ok=True)


# ============================================================
# MONTE CARLO SAMPLE CONFIGURATION
# ============================================================

SAMPLES = {
    "ttbar": {
        "file": os.path.join(
            MC_DIR,
            "spin_corr_ttbar_500k.root",
        ),
        "label": r"\textbf{$pp\to t\bar{t}$}",
        "color": "indigo",
        "linestyle": "-",
    },

    "eta_j_full": {
        "file": os.path.join(
            MC_DIR,
            "spin_corr_topponium_full_500k.root",
        ),
        "label": (
            r"\textbf{Simplified EFT $\eta_j$ full}"
        ),
        "color": "red",
        "linestyle": "--",
    },

    "eta_j_restrict": {
        "file": os.path.join(
            MC_DIR,
            "spin_corr_topponium_res_500k.root",
        ),
        "label": (
            r"\textbf{Simplified EFT $\eta_j$ restrict}"
        ),
        "color": "orange",
        "linestyle": "-",
    },
}


# ============================================================
# SPIN-CORRELATION OBSERVABLE CONFIGURATION
#
# error_mode:
#
#   "stat"
#       Use only Hist1D_y1_e1.
#
#   "stat+syst"
#       Combine Hist1D_y1_e1 and Hist1D_y1_e2:
#
#           sigma_total = sqrt(
#               sigma_stat^2 + sigma_syst^2
#           )
# ============================================================

OBSERVABLES = {
    # ========================================================
    # AZIMUTHAL SEPARATION
    # ========================================================

    "dphi": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "_Deltaphi_{ellell}_.root"
        ),
        "mc_hist": "h_dphi",

        "xlabel": (
            r"$|\Delta\phi_{\ell\ell}|$"
        ),

        "ylabel_variable": (
            r"|\Delta\phi_{\ell\ell}|"
        ),

        "xlim": (0.0, np.pi),
        "main_ylim": (0.0, 0.75),
        "ratio_ylim": (0.0, 3.2),

        "error_mode": "stat",
        "mean_precision": 2,

        "output": "01_spin_correlation_dphi_ll",
    },

    # ========================================================
    # LEPTON OPENING ANGLE IN THE LABORATORY FRAME
    # ========================================================

    "cosphi_lab": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "cosvarphi_{mathrm{lab}}.root"
        ),
        "mc_hist": "h_cos_opening_angle_lab",

        "xlabel": (
            r"$\cos\phi_{\mathrm{lab}}$"
        ),

        "ylabel_variable": (
            r"\cos\phi_{\mathrm{lab}}"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.0, 1.85),
        "ratio_ylim": (0.0, 5.0),

        "error_mode": "stat",
        "mean_precision": 2,

        "output": "02_spin_correlation_cosphi_lab",
    },

    # ========================================================
    # HELICITY-BASIS POLARIZATION OBSERVABLES
    # ========================================================

    "costheta_2_k": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{2}^{k}.root"
        ),
        "mc_hist": "h_lep_costheta_minus",

        "xlabel": (
            r"$\cos\!\left(\theta_{2}^{k}\right)$"
        ),

        "ylabel_variable": (
            r"\cos\!\left(\theta_{2}^{k}\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.44, 0.59),
        "ratio_ylim": (0.90, 1.08),

        "error_mode": "stat+syst",
        "mean_precision": 4,

        "output": "03_spin_correlation_costheta_2_k",
    },

    "costheta_1_k": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{1}^{k}.root"
        ),
        "mc_hist": "h_lep_costheta_plus",

        "xlabel": (
            r"$\cos\!\left(\theta_{1}^{k}\right)$"
        ),

        "ylabel_variable": (
            r"\cos\!\left(\theta_{1}^{k}\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.44, 0.59),
        "ratio_ylim": (0.90, 1.08),

        "error_mode": "stat+syst",
        "mean_precision": 4,

        "output": "04_spin_correlation_costheta_1_k",
    },

    # ========================================================
    # NORMAL-BASIS POLARIZATION OBSERVABLES
    # ========================================================

    "costheta_2_n": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{2}^{n}.root"
        ),
        "mc_hist": "h_lep_costheta_n_minus",

        "xlabel": (
            r"$\cos\!\left(\theta_{2}^{n}\right)$"
        ),

        "ylabel_variable": (
            r"\cos\!\left(\theta_{2}^{n}\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.44, 0.59),
        "ratio_ylim": (0.94, 1.06),

        "error_mode": "stat+syst",
        "mean_precision": 4,

        "output": "05_spin_correlation_costheta_2_n",
    },

    "costheta_1_n": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{1}^{n}.root"
        ),
        "mc_hist": "h_lep_costheta_n_plus",

        "xlabel": (
            r"$\cos\!\left(\theta_{1}^{n}\right)$"
        ),

        "ylabel_variable": (
            r"\cos\!\left(\theta_{1}^{n}\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.44, 0.59),
        "ratio_ylim": (0.94, 1.06),

        "error_mode": "stat+syst",
        "mean_precision": 4,

        "output": "06_spin_correlation_costheta_1_n",
    },

    # ========================================================
    # TRANSVERSE-BASIS POLARIZATION OBSERVABLES
    # ========================================================

    "costheta_2_r": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{2}^{r}.root"
        ),
        "mc_hist": "h_lep_costheta_r_minus",

        "xlabel": (
            r"$\cos\!\left(\theta_{2}^{r}\right)$"
        ),

        "ylabel_variable": (
            r"\cos\!\left(\theta_{2}^{r}\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.44, 0.59),
        "ratio_ylim": (0.94, 1.06),

        "error_mode": "stat+syst",
        "mean_precision": 4,

        "output": "07_spin_correlation_costheta_2_r",
    },

    "costheta_1_r": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{1}^{r}.root"
        ),
        "mc_hist": "h_lep_costheta_r_plus",

        "xlabel": (
            r"$\cos\!\left(\theta_{1}^{r}\right)$"
        ),

        "ylabel_variable": (
            r"\cos\!\left(\theta_{1}^{r}\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.44, 0.59),
        "ratio_ylim": (0.94, 1.06),

        "error_mode": "stat+syst",
        "mean_precision": 4,

        "output": "08_spin_correlation_costheta_1_r",
    },

    # ========================================================
    # HELICITY-BASIS SPIN-CORRELATION PRODUCT
    # ========================================================

    "c_kk": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{1}^{k}costheta_{2}^{k}.root"
        ),
        "mc_hist": "h_c_kk",

        "xlabel": (
            r"$\cos\theta_{1}^{k}"
            r"\cos\theta_{2}^{k}$"
        ),

        "ylabel_variable": (
            r"\left("
            r"\cos\theta_{1}^{k}"
            r"\cos\theta_{2}^{k}"
            r"\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.0, 1.70),
        "ratio_ylim": (0.0, 6.0),

        "error_mode": "stat+syst",
        "mean_precision": 3,

        "output": "09_spin_correlation_c_kk",
    },

    # ========================================================
    # NORMAL-BASIS SPIN-CORRELATION PRODUCT
    # ========================================================

    "c_nn": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{1}^{n}costheta_{2}^{n}.root"
        ),
        "mc_hist": "h_c_nn",

        "xlabel": (
            r"$\cos\theta_{1}^{n}"
            r"\cos\theta_{2}^{n}$"
        ),

        "ylabel_variable": (
            r"\left("
            r"\cos\theta_{1}^{n}"
            r"\cos\theta_{2}^{n}"
            r"\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.0, 1.70),
        "ratio_ylim": (0.0, 6.0),

        "error_mode": "stat+syst",
        "mean_precision": 3,

        "output": "10_spin_correlation_c_nn",
    },

    # ========================================================
    # TRANSVERSE-BASIS SPIN-CORRELATION PRODUCT
    # ========================================================

    "c_rr": {
        "data_file": (
            "HEPData-ins1742786-v1-"
            "costheta_{1}^{r}costheta_{2}^{r}.root"
        ),
        "mc_hist": "h_c_rr",

        "xlabel": (
            r"$\cos\theta_{1}^{r}"
            r"\cos\theta_{2}^{r}$"
        ),

        "ylabel_variable": (
            r"\left("
            r"\cos\theta_{1}^{r}"
            r"\cos\theta_{2}^{r}"
            r"\right)"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.0, 1.70),
        "ratio_ylim": (0.0, 6.0),

        "error_mode": "stat+syst",
        "mean_precision": 3,

        "output": "11_spin_correlation_c_rr",
    },

    # ========================================================
    # LEPTON OPENING ANGLE IN THE TTBAR FRAME
    # ========================================================

    "cosvarphi": {
        "data_file": (
            "HEPData-ins1742786-v1-cosvarphi.root"
        ),
        "mc_hist": "h_cos_opening_angle",

        "xlabel": (
            r"$\cos\phi$"
        ),

        "ylabel_variable": (
            r"\cos\phi"
        ),

        "xlim": (-1.0, 1.0),
        "main_ylim": (0.0, 1.70),
        "ratio_ylim": (0.0, 6.0),

        "error_mode": "stat+syst",
        "mean_precision": 3,

        "output": "12_spin_correlation_cosvarphi",
    },
}


# ============================================================
# INPUT FILE VALIDATION
# ============================================================

def validate_input_files():
    """
    Verify that every required CMS and MC ROOT file exists.
    """

    missing_files = []

    for configuration in SAMPLES.values():
        path = configuration["file"]

        if not os.path.isfile(path):
            missing_files.append(path)

    for configuration in OBSERVABLES.values():
        path = os.path.join(
            DATA_DIR,
            configuration["data_file"],
        )

        if not os.path.isfile(path):
            missing_files.append(path)

    if missing_files:
        message = (
            "The following required ROOT files were not found:\n"
            + "\n".join(
                f"  {path}"
                for path in missing_files
            )
        )

        raise FileNotFoundError(message)


# ============================================================
# READ CMS HEPDATA HISTOGRAM
# ============================================================

def read_hepdata(root_path, error_mode):
    """
    Read one normalized CMS HEPData distribution.

    Returns
    -------
    centers
        Bin centers.

    xerr
        Half-width of every bin.

    edges
        Histogram bin edges.

    y
        Central values.

    yerr
        Statistical uncertainty or total statistical plus
        systematic uncertainty.
    """

    if not os.path.isfile(root_path):
        raise FileNotFoundError(
            f"HEPData ROOT file not found:\n{root_path}"
        )

    with uproot.open(root_path) as root_file:
        top_level_keys = [
            key.split(";")[0]
            for key in root_file.keys()
        ]

        if not top_level_keys:
            raise RuntimeError(
                f"No ROOT objects were found in:\n{root_path}"
            )

        base = top_level_keys[0]

        central_name = f"{base}/Hist1D_y1"
        stat_name = f"{base}/Hist1D_y1_e1"
        syst_name = f"{base}/Hist1D_y1_e2"

        try:
            y, edges = root_file[
                central_name
            ].to_numpy()

            stat, _ = root_file[
                stat_name
            ].to_numpy()

        except KeyError as error:
            raise RuntimeError(
                f"Required HEPData objects were not found in:\n"
                f"{root_path}\n"
                f"Expected:\n"
                f"  {central_name}\n"
                f"  {stat_name}"
            ) from error

        if error_mode == "stat":
            yerr = stat

        elif error_mode == "stat+syst":
            try:
                syst, _ = root_file[
                    syst_name
                ].to_numpy()

            except KeyError as error:
                raise RuntimeError(
                    f"Systematic uncertainty object was not "
                    f"found in:\n{root_path}\n"
                    f"Expected:\n  {syst_name}"
                ) from error

            yerr = np.sqrt(
                stat**2 + syst**2
            )

        else:
            raise ValueError(
                f"Unknown uncertainty mode: {error_mode}"
            )

    y = np.asarray(
        y,
        dtype=np.float64,
    )

    yerr = np.asarray(
        yerr,
        dtype=np.float64,
    )

    edges = np.asarray(
        edges,
        dtype=np.float64,
    )

    centers = 0.5 * (
        edges[:-1] + edges[1:]
    )

    xerr = 0.5 * np.diff(edges)

    return centers, xerr, edges, y, yerr


# ============================================================
# NORMALIZE MONTE CARLO HISTOGRAM
# ============================================================

def normalize_density(values, edges):
    """
    Normalize a binned distribution such that

        sum_i f_i Delta x_i = 1.
    """

    values = np.asarray(
        values,
        dtype=np.float64,
    )

    edges = np.asarray(
        edges,
        dtype=np.float64,
    )

    bin_widths = np.diff(edges)

    area = np.sum(
        values * bin_widths
    )

    if not np.isfinite(area):
        raise RuntimeError(
            "The histogram normalization is not finite."
        )

    if area <= 0.0:
        raise RuntimeError(
            "The histogram normalization is zero or negative."
        )

    normalized_values = values / area

    return normalized_values


# ============================================================
# READ MONTE CARLO HISTOGRAM
# ============================================================

def read_mc_density(
    root_path,
    histogram_name,
    target_edges,
):
    """
    Read and normalize one MC histogram.

    If its binning differs from the CMS HEPData binning, the MC
    density is evaluated at the CMS bin centers using linear
    interpolation, reproducing the behavior of the original code.
    """

    if not os.path.isfile(root_path):
        raise FileNotFoundError(
            f"MC ROOT file not found:\n{root_path}"
        )

    with uproot.open(root_path) as root_file:
        available_keys = [
            key.split(";")[0]
            for key in root_file.keys()
        ]

        if histogram_name not in available_keys:
            available_text = "\n".join(
                f"  {key}"
                for key in available_keys
            )

            raise KeyError(
                f"Histogram '{histogram_name}' was not found in:\n"
                f"{root_path}\n\n"
                f"Available ROOT objects:\n"
                f"{available_text}"
            )

        values, edges = root_file[
            histogram_name
        ].to_numpy()

    values = np.asarray(
        values,
        dtype=np.float64,
    )

    edges = np.asarray(
        edges,
        dtype=np.float64,
    )

    target_edges = np.asarray(
        target_edges,
        dtype=np.float64,
    )

    values = normalize_density(
        values,
        edges,
    )

    same_binning = (
        len(edges) == len(target_edges)
        and np.allclose(
            edges,
            target_edges,
        )
    )

    if not same_binning:
        source_centers = 0.5 * (
            edges[:-1] + edges[1:]
        )

        target_centers = 0.5 * (
            target_edges[:-1]
            + target_edges[1:]
        )

        values = np.interp(
            target_centers,
            source_centers,
            values,
        )

    return values


# ============================================================
# DENSITY MEAN AND STANDARD DEVIATION
# ============================================================

def density_mean_std(values, edges):
    """
    Calculate the binned mean and standard deviation:

        mu = sum_i p_i x_i

        sigma = sqrt[
            sum_i p_i (x_i - mu)^2
        ]

    where

        p_i = f_i Delta x_i.
    """

    values = np.asarray(
        values,
        dtype=np.float64,
    )

    edges = np.asarray(
        edges,
        dtype=np.float64,
    )

    centers = 0.5 * (
        edges[:-1] + edges[1:]
    )

    widths = np.diff(edges)

    probability = values * widths
    normalization = np.sum(probability)

    if normalization <= 0.0:
        return np.nan, np.nan

    probability = (
        probability / normalization
    )

    mean = np.sum(
        probability * centers
    )

    variance = np.sum(
        probability
        * (centers - mean)**2
    )

    standard_deviation = np.sqrt(
        max(variance, 0.0)
    )

    return mean, standard_deviation


# ============================================================
# BUILD DIFFERENTIAL Y-AXIS LABEL
# ============================================================

def build_ylabel(variable_latex):
    """
    Construct

        (1/sigma) d sigma / dX

    using the observable-specific LaTeX expression X.
    """

    return (
        r"$\frac{1}{\sigma}"
        r"\frac{\mathrm{d}\sigma}"
        r"{\mathrm{d}"
        + variable_latex
        + r"}$"
    )


# ============================================================
# PLOT ONE SPIN-CORRELATION OBSERVABLE
# ============================================================

def plot_observable(
    observable_name,
    configuration,
):
    """
    Produce one normalized CMS-data/MC comparison with a
    Data/MC ratio panel.
    """

    data_path = os.path.join(
        DATA_DIR,
        configuration["data_file"],
    )

    (
        x,
        xerr,
        edges,
        y_data,
        yerr_data,
    ) = read_hepdata(
        root_path=data_path,
        error_mode=configuration["error_mode"],
    )

    # ========================================================
    # FIGURE AND PANEL CONFIGURATION
    # ========================================================

    fig = plt.figure(
        figsize=FIGSIZE,
        dpi=DPI,
        constrained_layout=False,
    )

    fig.set_size_inches(
        FIGSIZE[0],
        FIGSIZE[1],
        forward=True,
    )

    grid = GridSpec(
        nrows=2,
        ncols=1,
        height_ratios=[3.2, 1.0],
        hspace=0.05,
        figure=fig,
    )

    ax = fig.add_subplot(
        grid[0]
    )

    ratio_ax = fig.add_subplot(
        grid[1],
        sharex=ax,
    )

    legend_handles = []
    legend_labels = []

    # ========================================================
    # CMS DATA
    # ========================================================

    ax.errorbar(
        x,
        y_data,
        xerr=xerr,
        yerr=yerr_data,
        fmt="o",
        color="black",
        markerfacecolor="black",
        markeredgecolor="black",
        markersize=DATA_MARKER_SIZE,
        elinewidth=ERROR_LINE_WIDTH,
        capsize=CAP_SIZE,
        linewidth=ERROR_LINE_WIDTH,
        zorder=20,
    )

    cms_data_proxy = Line2D(
        [0],
        [0],
        color="black",
        marker="o",
        linestyle="none",
        markerfacecolor="black",
        markeredgecolor="black",
        markersize=DATA_MARKER_SIZE,
    )

    legend_handles.append(
        cms_data_proxy
    )

    legend_labels.append(
        r"\textbf{CMS Data}"
    )

    # ========================================================
    # MONTE CARLO DISTRIBUTIONS
    # ========================================================

    for sample_name, sample_configuration in SAMPLES.items():
        y_mc = read_mc_density(
            root_path=sample_configuration["file"],
            histogram_name=configuration["mc_hist"],
            target_edges=edges,
        )


        ax.stairs(
            y_mc,
            edges,
            linewidth=LINE_WIDTH,
            color=sample_configuration["color"],
            linestyle=sample_configuration["linestyle"],
            zorder=5,
        )

        # ====================================================
        # DATA / MC RATIO
        # ====================================================

        ratio = np.divide(
            y_data,
            y_mc,
            out=np.full_like(
                y_data,
                np.nan,
                dtype=np.float64,
            ),
            where=(y_mc > 0.0),
        )

        ratio_uncertainty = np.divide(
            yerr_data,
            y_mc,
            out=np.full_like(
                yerr_data,
                np.nan,
                dtype=np.float64,
            ),
            where=(y_mc > 0.0),
        )

        ratio_ax.errorbar(
            x,
            ratio,
            xerr=xerr,
            yerr=ratio_uncertainty,
            fmt="o",
            color=sample_configuration["color"],
            markerfacecolor=sample_configuration["color"],
            markeredgecolor=sample_configuration["color"],
            markersize=RATIO_MARKER_SIZE,
            elinewidth=ERROR_LINE_WIDTH,
            capsize=CAP_SIZE,
            linewidth=ERROR_LINE_WIDTH,
            zorder=10,
        )

        # ====================================================
        # LEGEND
        # ====================================================

        line_proxy = Line2D(
            [0],
            [0],
            color=sample_configuration["color"],
            linestyle=sample_configuration["linestyle"],
            linewidth=LINE_WIDTH,
        )


        legend_label = sample_configuration["label"]
        
        legend_handles.append(
            line_proxy
        )
        
        legend_labels.append(
            legend_label
        )
    

    # ========================================================
    # TOP TEXT
    # ========================================================

    ax.text(
        0.0,
        1.005,
        r"\textbf{Generator Level}",
        transform=ax.transAxes,
        fontsize=TITLE_SIZE,
        ha="left",
        va="bottom",
        clip_on=False,
    )

    ax.text(
        1.0,
        1.005,
        r"pp $\sqrt{s}=13$ TeV",
        transform=ax.transAxes,
        fontsize=TITLE_SIZE - 2,
        ha="right",
        va="bottom",
        clip_on=False,
    )

    # ========================================================
    # MAIN PANEL AXES
    # ========================================================

    ax.set_xlim(
        configuration["xlim"][0],
        configuration["xlim"][1],
    )

    ax.set_ylim(
        configuration["main_ylim"][0],
        configuration["main_ylim"][1],
    )

    ax.set_ylabel(
        build_ylabel(
            configuration["ylabel_variable"]
        ),
        fontsize=YLABEL_SIZE,
    )

    ax.tick_params(
        axis="both",
        which="major",
        labelsize=TICK_SIZE,
        direction="in",
        top=True,
        right=True,
    )

    ax.tick_params(
        axis="both",
        which="minor",
        labelsize=TICK_SIZE,
        direction="in",
        top=True,
        right=True,
    )

    plt.setp(
        ax.get_xticklabels(),
        visible=False,
    )

    # ========================================================
    # LEGEND
    # ========================================================

    ax.legend(
        legend_handles,
        legend_labels,
        fontsize=LEGEND_SIZE,
        loc="best",
        frameon=False,
        ncols=2,
        handlelength=2.0,
        handletextpad=0.8,
        labelspacing=0.8,
        borderaxespad=0.4,
    )

    # ========================================================
    # RATIO PANEL
    # ========================================================

    ratio_ax.axhline(
        1.0,
        color="black",
        linestyle="--",
        linewidth=1.3,
        zorder=1,
    )

    ratio_ax.set_xlim(
        configuration["xlim"][0],
        configuration["xlim"][1],
    )

    ratio_ax.set_ylim(
        configuration["ratio_ylim"][0],
        configuration["ratio_ylim"][1],
    )

    ratio_ax.set_xlabel(
        configuration["xlabel"],
        fontsize=LABEL_SIZE,
    )

    ratio_ax.set_ylabel(
        r"\textbf{Data/MC}",
        fontsize=RATIO_LABEL_SIZE,
    )

    ratio_ax.tick_params(
        axis="both",
        which="major",
        labelsize=TICK_SIZE,
        direction="in",
        top=True,
        right=True,
    )

    ratio_ax.tick_params(
        axis="both",
        which="minor",
        labelsize=TICK_SIZE,
        direction="in",
        top=True,
        right=True,
    )

    # ========================================================
    # LAYOUT
    # ========================================================

    fig.subplots_adjust(
        left=0.22,
        right=0.97,
        bottom=0.14,
        top=0.88,
        hspace=0.05,
    )

    # ========================================================
    # OUTPUT FILES
    # ========================================================

    output_pdf = os.path.join(
        OUTDIR,
        configuration["output"] + ".pdf",
    )

    output_png = os.path.join(
        OUTDIR,
        configuration["output"] + ".png",
    )

    fig.savefig(
        output_pdf,
        facecolor="white",
    )

    fig.savefig(
        output_png,
        dpi=DPI,
        facecolor="white",
    )

    print("\n" + "=" * 90)
    print(f"Observable : {observable_name}")
    print(f"CMS file   : {data_path}")
    print(f"MC object  : {configuration['mc_hist']}")
    print(f"PDF output : {output_pdf}")
    print(f"PNG output : {output_png}")

    plt.show()
    plt.close(fig)


# ============================================================
# VALIDATE ALL INPUTS
# ============================================================

validate_input_files()


# ============================================================
# PRODUCE ALL SPIN-CORRELATION FIGURES
# ============================================================

print("\n" + "=" * 90)
print("Producing all spin-correlation figures")
print("=" * 90)

for observable_name, configuration in OBSERVABLES.items():
    plot_observable(
        observable_name=observable_name,
        configuration=configuration,
    )


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 90)
print(
    f"All {len(OBSERVABLES)} spin-correlation "
    "figures were produced."
)
print(f"Output directory: {OUTDIR}")
print("=" * 90)
```
